# Exploratory Data Analysis - Bike Trips

## 1. Objective
The objective of this notebook is to consolidate and explore the bike trip data initially distributed across two separate files. This first step aims to verify the integrity, structure, and cleanliness of the dataset before proceeding with any in-depth analysis.


## 2. Data Loading and Merging
The raw data was provided in two parts:
* `bike_trips_part1.csv`
* `bike_trips_part2.csv`

**Action:** Concatenation of the two files.
**Result:** The final consolidated dataset contains **863,017 rows**.

```python
import pandas as pd

# 1. Read the two CSV files
df1 = pd.read_csv("../data/raw/bike_trips_part1.csv")
df2 = pd.read_csv("../data/raw/bike_trips_part2.csv")

# 2. Merge them (stack df2 below df1)
df = pd.concat([df1, df2], ignore_index=True)

# 3. Check the results
print(f"Rows in part 1: {len(df1)}")
print(f"Rows in part 2: {len(df2)}")
print(f"Total rows after merge: {len(df)}")

# Display the first few rows
display(df.head())
```

## 3. Dataset Structure
The merged dataset consists of **11 columns**. Here are the available variables and their data types:

* **Identifiers and Bikes:**
  * `Number` (int): Unique trip identifier.
  * `Bike number` (int): Bike identification number.
  * `Bike model` (str): Model of the bike used.
* **Temporal Information:**
  * `Start date` (str): Trip start date and time.
  * `End date` (str): Trip end date and time.
  * `Total duration` (str): Formatted trip duration.
  * `Total duration (ms)` (int): Total duration in milliseconds.
* **Spatial Information (Stations):**
  * `Start station number` (int) & `Start station` (str): Starting station number and name.
  * `End station number` (int) & `End station` (str): Ending station number and name.

*Note: The date columns (`Start date`, `End date`) are currently string types and will need to be converted to `datetime` format for temporal analysis.*

```python
# Check dataset shape (number of rows and columns)
print(f"Dataset shape: {df.shape}")

# List all columns
print("\nColumns list:")
print(df.columns.tolist())

# Check data types for each column
print("\nData types:")
print(df.dtypes)
```

## 4. Data Quality
An initial check of the data cleanliness revealed excellent overall quality:
* **Missing values:** The dataset is 100% complete. There are **no null values** (0.0%) across all columns.
* **Duplicates:** **No duplicated rows** were detected in the merged dataset.

```python
# Count missing values per column
missing_count = df.isna().sum()

# Calculate percentage of missing values
missing_percent = df.isna().mean() * 100

# Create a report DataFrame
missing_report = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percent": missing_percent
})

# Sort by highest missing percentage
print("Missing Values Report:")
display(missing_report.sort_values("missing_percent", ascending=False))
```

```python
# Count exact duplicate rows across all columns
duplicates_count = df.duplicated().sum()
print(f"Exact duplicate rows: {duplicates_count}")
```

```python
# Generate descriptive statistics for all columns (numerical and categorical)
print("Dataset Statistical Summary:")
display(df.describe(include="all").T)
```
